### **Rough Sets Theory**

**Rough Sets** are a mathematical theory for dealing with incomplete or imprecise data, proposed by Zdzisław Pawlak in the 1980s. Unlike other approaches, such as fuzzy logic, rough set theory does not require additional information, such as probability distributions or membership degrees, to describe imprecision. Instead, it is based on the idea that every object in a universe of discourse is associated with some information (data, knowledge).

Rough set theory uses the concept of **indiscernibility** to model knowledge. Two objects are indiscernible if they have the same values for a given set of attributes. This indiscernibility relation is an equivalence relation that partitions the universe of objects into equivalence classes.

For a set of objects $U$ and a set of attributes $A$, the indiscernibility relation $IND(A)$ partitions $U$ into equivalence classes. If we cannot distinguish between two objects using the attributes in $A$, they belong to the same indiscernibility class. A **rough set** is a representation of a set that cannot be precisely defined using only the existing equivalence classes. It is characterized by two exact sets:



1.  **Lower Approximation:** Contains all objects that *certainly* belong to the set, i.e., their equivalence classes are entirely contained within the set.

    $\underline{A}X = \{x \in U \mid [x]_A \subseteq X\}$

2.  **Upper Approximation:** Contains all objects that *possibly* belong to the set, i.e., their equivalence classes have a non-empty intersection with the set.

    $\overline{A}X = \{x \in U \mid [x]_A \cap X \neq \emptyset\}$

The **boundary region** of the set is the difference between the upper and lower approximations: $BN_A(X) = \overline{A}X - \underline{A}X$. The larger the boundary region, the "rougher" the set, indicating greater uncertainty about its membership.

### Discernibility Matrix

The **discernibility matrix** is a crucial tool for finding **reducts** and **cores** in datasets. It is used to identify which attributes can distinguish one object from another.

Given an information table with $n$ objects (rows) and $m$ attributes (columns), the discernibility matrix $M$ is an $n \times n$ matrix where each element $M_{ij}$ contains the set of attributes that distinguish object $x_i$ from object $x_j$.

$M_{ij} = \{a \in A \mid a(x_i) \neq a(x_j)\}$

If $M_{ij} = \emptyset$, then $x_i$ and $x_j$ are indiscernible with respect to all attributes in $A$.

From the discernibility matrix, the **discernibility function** can be constructed, which is a boolean expression used to find the reducts. A **reduct** is a minimal subset of attributes that preserves the ability to discern objects, i.e., it has the same indiscernibility relation as the complete set of attributes.

```python
#Import Libs
import pandas as pd
import numpy as np
```

```python
# Create a Function to Calc Reducts ( Simple Way )
def compute_reducts(df, target_column):

    # Separar atributos e classes
    attr = df.drop(target_column, axis=1)
    clss = df[target_column]

    # Inicialmente, assumimos que todos os atributos são relevantes
    reducts = set(attr.columns)

    # Iterar sobre os atributos para determinar relevância
    for col in attr.columns:
        temp_df = attr.drop(col, axis=1)
        temp_df['Class'] = clss

        if len(temp_df.groupby(list(temp_df.columns)).groups) < len(df):
            # Se a cardinalidade dos grupos não muda, o atributo é redundante
            reducts.discard(col)

    return reducts
```

```python
#Import Database
df = pd.read_csv('https://raw.githubusercontent.com/jrafa1607/DataScience-ML-AI/main/-%20Datasets/diabetes.csv')
```

```python
#Replace the Last Column Name ( Y ) for Class
df.rename(columns={'Outcome': 'Class'}, inplace=True)
```

```python
#Show the Results after Change the Column Name
print(df)
```

```python
# Calcular reducts
reducts = compute_reducts(df, 'Class')
```

```python
#Show the Results
print("Most Relevants Attributes:", reducts)
```
