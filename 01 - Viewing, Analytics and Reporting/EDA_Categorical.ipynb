```python
#Import Libs
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
#Ignore Error Message
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
```

```python
#Import Database for EDA
df = pd.read_csv('https://raw.githubusercontent.com/jrafa1607/DataScience-ML-AI/main/-%20Datasets/census.csv')
```

```python
#Database Visualization
print(df)
```

```python
#Show the Most Frequent Value in the Attribute
print(df['age'].mode())
print(df['workclass'].mode())
```

```python
#Examples of Graphical Ilustrations for Categorical Data
```

```python
graph = px.treemap(df, path=['workclass', 'age'])
graph.show()
```

```python
graph = px.treemap(df, path=['occupation', 'relationship', 'age'])
graph.show()
```

```python
graph = px.parallel_categories(df, dimensions=['occupation', 'relationship'])
graph.show()
```

```python
graph = px.parallel_categories(df, dimensions=['workclass', 'occupation', 'income'])
graph.show()
```

```python
graph = px.parallel_categories(df, dimensions=['education', 'income'])
graph.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(y='workclass', data=df, order = df['workclass'].value_counts().index)
plt.title('Distribution of Workclass')
plt.xlabel('Count')
plt.ylabel('Workclass')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='sex', data=df)
plt.title('Distribution of Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(12, 7))
sns.countplot(y='education', hue='income', data=df, order = df['education'].value_counts().index)
plt.title('Income Distribution by Education Level')
plt.xlabel('Count')
plt.ylabel('Education')
plt.show()
```

```python
plt.figure(figsize=(15, 10))
crosstab_df = pd.crosstab(df['occupation'], df['education'])
sns.heatmap(crosstab_df, cmap='YlGnBu', fmt='d', annot=False)
plt.title('Relationship between Occupation and Education')
plt.xlabel('Education')
plt.ylabel('Occupation')
plt.show()
```

```python
plt.figure(figsize=(15, 10))
sns.countplot(y='native-country', data=df, order = df['native-country'].value_counts().index)
plt.title('Distribution of Native Country')
plt.xlabel('Count')
plt.ylabel('Native Country')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='age', hue='income', multiple='stack', kde=True)
plt.title('Age Distribution by Income')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```
