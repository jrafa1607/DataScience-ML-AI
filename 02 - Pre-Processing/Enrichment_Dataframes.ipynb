```python
#Import Libs
import pandas as pd
import numpy as np
import random
import string
```

```python
#Create the Dataframe
df_example = pd.DataFrame({'Data_Column': [pd.NA] * 10})
print(df_example)
```

```python
#Example 01 - Enrich with Random Integer Values (Between 1 and 101)
random_integers = np.random.randint(1, 101, size=10)
df_example['Data_Column'] = random_integers
print(df_example)
```

```python
#Example 02 - Enrich with Random Float Values (Between 0 and 1)
random_floats = np.random.uniform(0, 1, size=10)
df_example['Data_Column'] = random_floats
print(df_example)
```

```python
#Example 03 - Enrich with Random String Values (Text)
def generate_random_string(length=10):
    letters = string.ascii_lowercase
    return ''.join(random.choice(letters) for i in range(length))

random_strings = [generate_random_string() for _ in range(10)]
df_example['Data_Column'] = random_strings
print(df_example)
```

```python
#Example 04 - Enrich with Random Boolean Values (True/False)
random_booleans = np.random.choice([True, False], size=10)
df_example['Data_Column'] = random_booleans
print(df_example)
```

```python
#Example 05 - Enrich with Specific Values ( List Predefined )
list = ['Zero', 'One', 'Two', 'Three', 'Four', 'Five', 'Six', 'Seven', 'Eight', 'Nine']
df_example['Data_Column'] = list
print(df_example)
```

```python
#Example 06 - Enrich with Specific Random Values ( List Predefined )
list = ['Zero', 'One', 'Two']
random_selections = np.random.choice(list, size=len(df_example))
df_example['Data_Column'] = random_selections
print(df_example)
```
